```python
# Storing structured records using a list of dictionaries
orders = [
    {"id": 101, "category": "Books", "amount": 450},
    {"id": 102, "category": "Electronics", "amount": 2200},
    {"id": 103, "category": "Stationery", "amount": 320},
    {"id": 104, "category": "Books", "amount": 780},
    {"id": 105, "category": "Electronics", "amount": 1500},
]

print("Total orders stored:", len(orders))
print("First order record:", orders[0])
print("Second order category:", orders[1]["category"])
```

```python
# Function using a loop to calculate total and average sales
def summarize_sales(records):
    total = 0
    for row in records:
        total += row["amount"]
    average = total / len(records) if records else 0
    return total, average

total_sales, avg_sales = summarize_sales(orders)
print(f"Total Sales: {total_sales}")
print(f"Average Order Value: {avg_sales:.2f}")

# Filter high-value orders (>= 1000)
high_value_orders = [row for row in orders if row["amount"] >= 1000]
print("High-value orders count:", len(high_value_orders))
```
